```python
import os
import subprocess
from pathlib import Path

from docling.backend.docling_parse_backend import DoclingParseDocumentBackend
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import (
    EasyOcrOptions,
    OcrMacOptions,
    PdfPipelineOptions,
    RapidOcrOptions,
    TesseractCliOcrOptions,
    TesseractOcrOptions,
)
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling_core.types.doc.base import ImageRefMode

os.chdir("../")
```

```python
source = "data/SPE_NV_2025_EM_01_V2_FORGB_BIO_AL_IMP_CROPPED.pdf"  # PDF path or URL
source_dir = "data"
output_dir = 'data/processed'
```

```python
# %%
pipeline_options = PdfPipelineOptions()
pipeline_options.do_table_structure = True
pipeline_options.table_structure_options.do_cell_matching = True
pipeline_options.generate_picture_images = False
pipeline_options.do_formula_enrichment = True

pipeline_options.do_ocr = True
ocr_options = TesseractCliOcrOptions(force_full_page_ocr=True)
pipeline_options.ocr_options = ocr_options
```

```python
# %%
# !docling ./files --from docx --to md --output ./scratch
input_doc = source
converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_options=pipeline_options,
        )
    }
)

conv_res = converter.convert(input_doc)
```

```python
# Save markdown with embedded pictures
source_path = Path(source)  # Converte o caminho do source para um objeto Path
processed_path = source_path.parent / "processed" / (source_path.stem + ".md")  # Adiciona 'processed' no meio e '.md' no final

# Cria o diretório 'processed' se não existir
processed_path.parent.mkdir(parents=True, exist_ok=True)

# Salva o arquivo markdown no caminho ajustado
conv_res.document.save_as_markdown(str(processed_path))
```
