```python
# Caminho do arquivo Markdown
markdown_file = Path("data/processed/SPE_NV_2025_EM_01_V2_FORGB_BIO_AL_IMP_CROPPED.md")
```

```python
# Ler o conteúdo do arquivo Markdown
with open(markdown_file, "r", encoding="utf-8") as file:
    markdown_content = file.read()

# Dividir o conteúdo em chunks com base nos títulos e subtítulos
def split_markdown_by_headers(content):
    chunks = []
    current_chunk = []
    for line in content.splitlines():
        if "# " in line:  # Detecta títulos e subtítulos
            if current_chunk:
                chunks.append("\n".join(current_chunk))
                current_chunk = []
        current_chunk.append(line)
    if current_chunk:
        chunks.append("\n".join(current_chunk))
    return chunks

chunks = split_markdown_by_headers(markdown_content)

# Criar embeddings para os chunks
embeddings = OpenAIEmbeddings()  # Certifique-se de configurar sua chave de API da OpenAI
vectorstore = FAISS.from_texts(chunks, embeddings)

# Salvar o vectorstore localmente
vectorstore.save_local("data/processed/vectorstore")
```

```python
# Query vectorstore with a specific vector or text
query = "Expansão do universo"  # Replace with your query
results = vectorstore.similarity_search(query, k=1)  # Retrieve the top 1 match

# Print the result
for result in results:
    print("Matched text:", result.page_content)
```
